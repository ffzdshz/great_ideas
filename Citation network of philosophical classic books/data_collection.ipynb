```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from tqdm.notebook import tqdm
```

```python
book_phil = pd.read_excel('philbooklist.xlsx', header = 0)
```

```python
book_phil = book_phil[['Title','Author','Original_title']]
```

```python
book_phil['id'] = book_phil.index
```

```python
book_phil['id'] = pd.factorize(book_phil['id'])[0]
```

```python
book_phil.index = np.arange(len(book_phil.index))
```

```python
import requests as req
```

```python
import time
```

```python
df = pd.DataFrame(columns=['id1','author1','author2','get-title','get-authors','searchInfo'])
```

```python
dataset = pd.DataFrame(columns=['kind','totalItems','items'])
```

```python
proxies = {
  'http': 'http://127.0.0.1:8001',
  'https': 'http://127.0.0.1:8001',
}
```

```python
for i in range(len(book_phil)):
    t1 = book_phil[book_phil['id']==i]['Title'][i]
    a1 = book_phil[book_phil['id']==i]['Author'][i]
    book_phil1 = book_phil[book_phil.index <= i]
    namelsit1 = book_phil1.drop_duplicates(subset=['Title'], keep='first')
    namelsit1.index = np.arange(len(namelsit1.index))
    for j in range(len(namelsit1)):
        t2 = namelsit1['Title'][j]
        url = 'https://www.googleapis.com/books/v1/volumes?q='+'''"'''+t2+'''"'''+' intitle:'+'''"'''+t1+'''"'''+' inauthor:'+a1
        h = {'Authorization': 'AIzaSyBd3tDlB05G2o9VxNyV_r64casG9V3S4aM'}
        while True:
            try:
                resp = req.get(url, headers=h, proxies=proxies)    
                data = resp.json()
                if data['totalItems'] !=0:
                    dataset = dataset.append(pd.DataFrame(data))
                    data1 = data['items']
                    for k in range(len(data1)):
                        if 'searchInfo' in data1[k]:
                            list = [t1,a1,a2,data1[k]['volumeInfo'].get('title'),data1[k]['volumeInfo'].get('authors'),data1[k].get('searchInfo')]
                            df1 = pd.DataFrame(list).T
                            df1.columns = ['id1','author1','author2','get-title','get-authors','searchInfo']
                            df = df.append(df1)
                break
            except Exception:
                pass
```

```python
writer = pd.ExcelWriter('F:\\dataset\\great_book\\cite_phil.xlsx')
df.to_excel(writer)
writer.save()
```
